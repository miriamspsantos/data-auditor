## **Changing feature types**

```python
import sys
sys.path.append("../src")
```

```python
from profiler import DataProfiler
from preprocessing import DataPreprocessor
from utils import load_data
```

```python
df = load_data("../data/hcc_dataset.csv")
```

```python
df
```

```python
hcc_profile = DataProfiler(df)
```

```python
hcc_profile.print_report()
```

```python
hcc_processor = DataPreprocessor(hcc_profile)
```

```python
hcc_processor.print_log_report()
```

```python
hcc_processor.set_semantic_feature_types(
    ordinal_cols=["Age", "Symptoms"])
```

```python
hcc_profile.print_report()
```

```python
hcc_processor.set_semantic_feature_types(
    ordinal_cols=["Endemic"])
```

```python
hcc_profile.print_report()
```

```python
hcc_processor.print_log_report()
```

The `set_semantic_features` is supposed to be overriden. The changes made are lost between runs. The user can check what has been done in `print_log_report()`.

```python
hcc_profile.print_report()
```

```python
hcc_processor.change_feature_types(
    num_cols=["aht", "alcohol"],
    cat_cols=["afp", "age"]
)
```

```python
hcc_profile.print_report()
```
