```python
import pandas as pd
from sklearn.pipeline import Pipeline
from encoders.custom_transformer import CustomTransformer, MultiplyColumns
```

```python
df = pd.DataFrame({"a": [1, 2, 3], "b": [4, 5, 6], "c": [7, 8, 9]})
```

```python
df
```

```python
pipe = Pipeline(
    steps=[
        ("use_custom_transformer", CustomTransformer())
    ]
)

transformed_df = pipe.fit_transform(df)
```

```python
transformed_df
```

`pipe.fit_transform(df)` is a shorthand for `pipe.fit(df)` and `pipe.transform(df)`. Internally, this means:
- `Pipeline.fit(df)` calls `CustomTransformer.fit(df)`
- `Pipeline.transform(df)` calls `CustomTransformer.transform(df)`

It passes the data through each transformer step **in order** and returns the final transformed data.

- Each step **must implement** `fit()` and `transform()`
- `fit_transform()` is automatically handled if we inherit from `TransformerMixin`

### **What Makes a Custom Class Compatible?**
- It must inherit from `BaseEstimator` and `TransformerMixin` or at least implement `.fit(X, y=None)` and `.transform()`
- Otionally, it the last step is a model (e.g., Logistic Regression), it must implement `.fit(X, y)` and `.predict(X)` or `.predict_proba(X)`.
