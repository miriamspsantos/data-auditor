```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["HR", "IT", "IT", "HR", "Sales", "Sales", "IT"],
    "Employee": ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank", "Grace"],
    "Salary": [50000, 60000, 65000, 52000, 70000, 72000, 58000],
    "Gender": ["F", "M", "M", "F", "F", "M", "F"]
})
```

```python
df
```

```python
df.groupby("Department")["Employee"].count()
```

```python
df.groupby("Department").agg({
    "Salary": ["mean", "max", "min"],
    "Employee": "count"
})
```

```python
df.sort_values(by="Salary", ascending=False)
```

```python
df["DeptRank"] = df.groupby("Department")["Salary"].rank(ascending=False)
```

```python
df
```

```python
df.pivot_table(
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "class": ["NaN", "A", "B", "B", "C", "C"],
    "gender": ["F", "M", "F", "F", "M", "F"],
    "score": [90, None, 75, 85, 92, None]
})
```

```python
df
```

```python
pd.pivot_table(
    df,
    index="class",
    columns="gender",
    values="score",
    aggfunc="count",
    #fill_value=0
)
```

```python
pd.pivot_table(
    df,
    index="class",
    columns="gender",
    aggfunc="size"
)
```

- Use "count" → when you care about non-missing data in a specific column
- Use "size" → when you care about row frequency, regardless of missing data
