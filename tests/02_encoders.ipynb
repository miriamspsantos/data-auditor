## **Decoding OneHotEncoder (tests)**

```python
import pandas as pd
import numpy as np

X = pd.DataFrame({
    "OHE_A": [1, 0, 0],
    "OHE_B": [0, 1, 0]
})

def decode_row(row, ohe_list):
    for ohe_col, category in ohe_list:
        if ohe_col in row and row[ohe_col] == 1:
            return category
    return np.nan

ohe_list = [("OHE_A", "A"), ("OHE_B", "B")]

X["decoded"] = X.apply(decode_row, axis=1, args=(ohe_list,))
print(X)
```

- **What is row in `apply`?**
    - When we do `X.apply(decode_row, axis=1)`:
        - Each row is passed as a pandas Series
        - `row` is a Series where **index = column names** and **values = row values**
- **Why `if ohe_col in row` works:**
    - It checks if `ohe_col` (string name) is in the series

```python
X = pd.DataFrame({
    "OHE_A": [1, 0, 0],
    "OHE_B": [0, 1, 0]
})
```

```python
X
```

```python
ohe_cols = list(X.columns)
X.drop(columns=[col for col in ohe_cols if col in X.columns], inplace=True)
result = pd.DataFrame({
    "OHE_C": [1, 0, 0]})

X = pd.concat([X, result], axis=1)
```

## **Using Encoders in ColumnTransformer**

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [25, 40, 32, 29, 50],                          # Numerical
    "Education": ["High School", "PhD", "Bachelors", "Masters", "Bachelors"],  # Ordinal
    "City": ["New York", "Paris", "Tokyo", "Paris", "Tokyo"]  # Nominal
})

df
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler
from encoders.ordinal import ProfilerOrdinalEncoder
from encoders.onehot import ProfilerOneHotEncoder
```

```python
num_features = ["Age"]
ordinal_features = ["Education"]
cat_features = ["City"]
```

```python
preprocessor = ColumnTransformer([
    ("scale", MinMaxScaler(), num_features),
    ("encode_ordinal", ProfilerOrdinalEncoder(), ordinal_features),
    ("encode_one_hot", ProfilerOneHotEncoder(), num_features)
])
```

```python
df_encoded = preprocessor.fit_transform(df)
```

```python
df_encoded
```

## **Using Encoders in Pipelines**

```python
from sklearn.pipeline import Pipeline
```

```python
pipeline = Pipeline(steps=[
    ("preprocess", preprocessor)
])
```

```python
pipeline.fit_transform(df)
```
